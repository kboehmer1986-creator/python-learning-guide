### Basic Data Types

| Type      | Example          | Description                     |

## Basic Operations

```python
# Arithmetic operations
a = 10
b = 3

# Basic operations
print(f"Addition: {a + b}")        # 13
print(f"Subtraction: {a - b}")     # 7
print(f"Multiplication: {a * b}")  # 30
print(f"Division: {a / b}")        # 3.333...
print(f"Modulus: {a % b}")         # 1
print(f"Exponent: {a ** b}")       # 1000
print(f"Floor Division: {a // b}") # 3
```

```python
# Comparison operators
print(f"Equal: {a == b}")      # False
print(f"Not Equal: {a != b}")   # True
print(f"Greater Than: {a > b}") # True
print(f"Less Than: {a < b}")    # False
```

```python
# Logical operators
x = True
y = False

print(f"AND: {x and y}")   # False
print(f"OR: {x or y}")     # True
print(f"NOT: {not x}")     # False
```

## Control Flow

### Conditional Statements

```python
# If-elif-else example
age = 18
if age < 13:
    print("Child")
elif age < 18:
    print("Teenager")
else:
    print("Adult")
```

### Loops

```python
# For loop example
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# While loop example
count = 0
while count < 5:
    print(count)
    count += 1
```

## Functions

```python
# Function example
def greet(name):
    """This function greets the person passed in as a parameter"""
    return f"Hello, {name}!"

# Call the function
greet("Kay")
```

```python
# Function with default parameters
def greet_default(name="User"):
    """This function greets with a default name"""
    return f"Hello, {name}!"

greet_default()        # Uses default argument
greet_default("Alice") # Uses provided argument
```

## Data Structures

### Lists

```python
# List example
fruits = ["apple", "banana", "cherry"]
fruits.append("orange")    # Add to end
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'orange']
```

```python
# List comprehension
squares = [x**2 for x in range(10)]
squares
```

### Dictionaries

```python
# Dictionary example
person = {"name": "Kay", "age": 39}
print(person["name"])  # Output: Kay
```

```python
# Dictionary comprehension
square_dict = {x: x**2 for x in range(5)}
square_dict
```

## File Handling

```python
# Writing to a file
with open("example.txt", "w") as file:
    file.write("Hello, World!\nThis is a test file.")

# Reading from a file
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

## Error Handling

```python
# Try-except example
try:
    # Code that might raise an exception
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
except Exception as e:
    print(f"An error occurred: {e}")
else:
    print("No errors occurred")
finally:
    print("This always executes")
```

### Exercise 2: Control Flow
1. Write a program that checks if a number is positive, negative, or zero.
2. Write a for loop to print the first 10 even numbers.
3. Write a while loop to print numbers from 10 to 1.

```python
# Exercise 2 solution
# 1. Check number
num = float(input("Enter a number: "))
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")

# 2. First 10 even numbers
for i in range(2, 21, 2):
    print(i)

# 3. Countdown from 10 to 1
i = 10
while i >= 1:
    print(i)
    i -= 1
```
