## Decorators

Decorators allow you to modify the behavior of functions or classes.

```python
# Example: A simple decorator
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished {func.__name__}")
        return result
    return wrapper

@my_decorator
def greet(name):
    return f"Hello, {name}!"

greet("Kay")
```

### Decorator with Arguments

You can also pass arguments to decorators.

```python
def repeat(num_times):
    def decorator_repeat(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator_repeat

@repeat(num_times=3)
def say_hello(name):
    print(f"Hello, {name}!")

say_hello("Alice")
```

## Generators

Generators allow you to iterate over a sequence without loading it entirely into memory.

```python
# Example: A simple generator
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

# Use the generator
for num in count_up_to(5):
    print(num)
```

```python
# Example: Fibonacci generator
def fibonacci(limit):
    a, b = 0, 1
    while a <= limit:
        yield a
        a, b = b, a + b

# Print Fibonacci numbers up to 100
list(fibonacci(100))
```

## Context Managers

Context managers help manage resources like files or database connections.

```python
# Example: Custom context manager
class MyContextManager:
    def __enter__(self):
        print("Entering the context")
        return self
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Exiting the context")

# Use the context manager
with MyContextManager() as cm:
    print("Inside the context")
```

## Multithreading and Multiprocessing

Multithreading and multiprocessing allow you to run tasks concurrently.

```python
import threading
import time

# Example: Multithreading
def worker(thread_name, delay):
    for i in range(5):
        time.sleep(delay)
        print(f"Thread {thread_name}, iteration {i}")

# Create threads
thread1 = threading.Thread(target=worker, args=("A", 1))
thread2 = threading.Thread(target=worker, args=("B", 1.5))

# Start threads
thread1.start()
thread2.start()

# Wait for threads to complete
thread1.join()
thread2.join()

print("All threads completed")
```

## Metaclasses

Metaclasses allow you to control the creation of classes.

```python
# Example: Custom metaclass
class Meta(type):
    def __new__(cls, name, bases, dct):
        print(f"Creating class {name}")
        return super().__new__(cls, name, bases, dct)

class MyClass(metaclass=Meta):
    pass

# This will trigger the metaclass
obj = MyClass()
```

## Functional Programming

Python supports functional programming concepts like `map`, `filter`, and `reduce`.

```python
from functools import reduce

# Example: Using map, filter, and reduce
numbers = [1, 2, 3, 4, 5]

# Map: Square each number
squared = list(map(lambda x: x**2, numbers))
print("Squared:", squared)

# Filter: Get even numbers
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Evens:", evens)

# Reduce: Sum all numbers
total = reduce(lambda x, y: x + y, numbers)
print("Total:", total)
```

## Error Handling and Debugging

Python provides robust tools for error handling and debugging.

```python
# Example: Using try-except blocks
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Error: Division by zero!")
        return None

# Test the function
print(divide(10, 2))  # Valid division
print(divide(10, 0))  # Division by zero
```
