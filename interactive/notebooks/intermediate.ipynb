## Functions

Functions are reusable blocks of code that perform a specific task.

```python
# Example: Basic function
def greet(name):
    """This function greets the person passed in as a parameter"""
    return f"Hello, {name}!"

# Call the function
greet("Kay")
```

```python
# Example: Function with default arguments
def greet_default(name="User"):
    """This function greets with a default name"""
    return f"Hello, {name}!"

greet_default()        # Uses default argument
greet_default("Alice") # Uses provided argument
```

## File Handling

File handling allows you to read from and write to files.

```python
# Example: Writing to a file
with open("example.txt", "w") as file:
    file.write("Hello, World!\nThis is a test file.")

# Example: Reading from a file
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# Example: Reading line by line
with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())
```

## Error Handling

Error handling allows you to manage exceptions and errors gracefully.

```python
# Example: Using try-except blocks
try:
    # Code that might raise an exception
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
except Exception as e:
    print(f"An error occurred: {e}")
else:
    print("No errors occurred")
finally:
    print("This always executes")
```

## Modules and Packages

Modules and packages allow you to organize and reuse code.

```python
# Example: Importing a module
import math
print(math.sqrt(16))  # Output: 4.0

# Example: Importing specific functions
from math import pi
print(pi)  # Output: 3.141592653589793
```

## Data Structures Deep Dive

Let's explore more advanced uses of Python's data structures.

```python
# Example: List comprehension
squares = [x**2 for x in range(10)]
print("Squares:", squares)

# Example: Dictionary comprehension
square_dict = {x: x**2 for x in range(5)}
print("Square dictionary:", square_dict)

# Example: Set comprehension
unique_squares = {x**2 for x in [-2, -1, 0, 1, 2]}
print("Unique squares:", unique_squares)
```

```python
# Example: Working with lists
numbers = [1, 2, 3, 4, 5]

# Map: Square each number
squared = list(map(lambda x: x**2, numbers))
print("Squared:", squared)

# Filter: Get even numbers
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Evens:", evens)

# Reduce: Sum all numbers
from functools import reduce
total = reduce(lambda x, y: x + y, numbers)
print("Total:", total)
```
