## Object-Oriented Programming Solutions

### Solution 15: Basic Class

```python
# Solution for Exercise 15
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def greet(self):
        return f"Hello, I'm {self.name} and I'm {self.age} years old."

# Test the class
person = Person("Kay", 39)
print(person.greet())
```

### Solution 16: Inheritance

```python
# Solution for Exercise 16
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id
    
    def greet(self):
        return f"Hello, I'm {self.name}, I'm {self.age} years old, and my student ID is {self.student_id}."

# Test the class
student = Student("Alice", 20, "S12345")
print(student.greet())
```

## Advanced Topics Solutions

### Solution 17: Decorator

```python
# Solution for Exercise 17
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"{func.__name__} took {end_time - start_time:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(1)
    return "Done"

# Test the decorator
slow_function()
```

### Solution 18: Generator

```python
# Solution for Exercise 18
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

# Test the generator
list(fibonacci(10))
```

### Solution 19: Context Manager

```python
# Solution for Exercise 19
import time
from contextlib import contextmanager

@contextmanager
def Timer():
    start_time = time.time()
    yield
    end_time = time.time()
    print(f"Elapsed time: {end_time - start_time:.4f} seconds")

# Test the context manager
with Timer():
    time.sleep(1)
```

### Solution 20: Multithreading

```python
# Solution for Exercise 20
import threading

def print_numbers(start, end):
    for i in range(start, end + 1):
        print(i)

# Create threads
thread1 = threading.Thread(target=print_numbers, args=(1, 5))
thread2 = threading.Thread(target=print_numbers, args=(6, 10))

# Start threads
thread1.start()
thread2.start()

# Wait for threads to complete
thread1.join()
thread2.join()
print("All threads completed")
```
